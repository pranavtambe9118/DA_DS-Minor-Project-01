Q1.What is the difference between the `is` operator and the `==` operator? Give one example where
two variables return True for `==` but False for `is`.

Answer:
The '==' operator is used to check weather the two variables have same value or not whereas the 'is' operator check that weather the two given variables refer to an exact same object in the memory.


```python
#Example for Q1
a = [10, 20, 30]
b = [10, 20, 30]
#Here 'a' and 'b' are variables

print(a == b)
#"==" operator is used here and both lists contain the same values, so '==' returns True

print(a is b)
#"is" operator is used here and even though both list contain the same values they are two separate objects stored in memory, so 'is' returns False.
```

Q5. Why is dictionary lookup O(1) on average while searching for a value in a list using `in` is O(n)?
Explain in plain English without using the word "hash" - then explain it again WITH the word "hash".

Answer:

Without using the word "hash":

A dictionary can directly locate an item using its key, so it usually does not need to check every element. But in a list Python may have to check each element one by one until it finds the required value. That's why dictionary lookup is generally faster.

With the word "hash":

A dictionary uses a hash function to find the location of a particular key. This makes searching very fast, with an average time complexity of O(1). In a list, elements are checked one by one, so searching takes O(n) time in the worst case.

Q7. What is the difference between `list.append(x)`, `list.extend(x)`, and `list.insert(i, x)`? For each,
show what happens when `x = [4, 5]` is added to `a = [1, 2, 3]`.

Answer:

1. append()

It adds the complete item at the end of the list.

```python
a = [1, 2, 3]
a.append([4, 5])

print(a)
```

2. extend()

It adds each element of the given list separately.

```python
a = [1, 2, 3]
a.extend([4, 5])

print(a)
```

3. insert()

It adds an element at the position specified by the index.

```python
a = [1, 2, 3]
a.insert(2, [4, 5])

print(a)
```

The main difference is that append() adds one complete item, extend() adds elements separately and insert() adds an item at a particular position.

Q9. Explain the difference between `/`, `//`, `%`, and `**` operators in Python. For each, give the
output of the operation when applied to (-7, 2).

Answer:

These operators are used for different mathematical operations in Python.

1. '/' is used for Normal Division.

2. '//' is used for Floor Division.

3. '%' is used for Modulus(remainder).

4. '**' is used for Power.


```python
#When applied to (-7, 2).

print(-7 / 2)
print(-7 // 2)
print(-7 % 2)
print(-7 ** 2)
```

Explanation:
- / performs normal division.
- // gives the floor value of the division.
- % returns the remainder.
- ** is used to calculate the power of a number.

One important thing is that -7 ** 2 gives -49 because Python calculates the power before applying the negative sign. But (-7) ** 2 gives 49.

Q10. Why is NumPy significantly faster than Python lists for numerical operations? Give 3 reasons, at
least one should mention memory layout.

Answer:

NumPy is generally faster than normal Python lists when performing numerical calculations.

There are three main reasons for this:

1. Memory layout: NumPy stores elements of the same data type in a compact and continuous memory layout, which makes accessing data more efficient.

2. Vectorization: We can perform operations on an entire array directly instead of writing separate loops for every element.

3. Optimized code: NumPy uses highly optimized code written in C and other compiled languages, which helps in performing calculations faster
