Q11. Predict the output. Why does the second line behave differently from the first?

matrix1 = [[0] * 3 for _ in range(3)]

matrix2 = [[0] * 3] * 3
  
matrix1[0][0] = 99

matrix2[0][0] = 99
  
print(matrix1)

print(matrix2)



Answer:

- Predicted output:

[[99, 0, 0], [0, 0, 0], [0, 0, 0]]

[[99, 0, 0], [99, 0, 0], [99, 0, 0]]

```python
matrix1 = [[0] * 3 for _ in range(3)]
matrix2 = [[0] * 3] * 3

matrix1[0][0] = 99
matrix2[0][0] = 99

print(matrix1)
print(matrix2)
```

In matrix1, each row is created separately, so changing one row does not affect the others. But in matrix2, all three rows refer to the same list. That's why changing one element changes all three rows.

- Why does the second line behave differently from the first?

Reason:

- matrix1 creates three independent lists, so changing one row affects only that row.

- matrix2 repeats references to the same list three times, so changing one row affects all three rows.

Q12. Predict the output:

x = [1, 2, 3]

y = x

x = x + [4]

print("x =", x)

print("y =", y)
  

a = [1, 2, 3]

b = a

a += [4]

print("a =", a)

print("b =", b)

Now answer: Why do `x = x + [4]` and `a += [4]` behave differently?

Answer:

- Predicted output:

x = [1, 2, 3, 4]

y = [1, 2, 3]

a = [1, 2, 3, 4]

b = [1, 2, 3, 4]

```python
x = [1, 2, 3]
y = x

x = x + [4]

print("x =", x)
print("y =", y)

a = [1, 2, 3]
b = a

a += [4]

print("a =", a)
print("b =", b)
```

x = x + [4] creates a new list and assigns it to x, so y still refers to the original list. But a += [4] modifies the existing list itself, so b also shows the updated values.

- Why do `x = x + [4]` and `a += [4]` behave differently?

Reason:

- x = x + [4] creates a new list and makes x refer to it. y still refers to the original list.

- a += [4] modifies the existing list in place. Since b refers to that same list, its contents also change.

Q13. Predict the output. What is the value of `i` after the loop ends, and why?

for i in range(5):

    pass

print(i)
  

for j in range(0):

    pass

print(j)

Answer:

- Predicted output:

4

NameError: name 'j' is not defined

```python
for i in range(5):
    pass

print(i)

for j in range(0):
    pass

print(j)
```

The first loop runs with values from 0 to 4, so the final value of i is 4. The second loop does not run even once because range(0) is empty. Therefore, j is never created, and printing it gives a NameError.

Note: The first output prints successfully. The second print statement raises an error.

- What is the value of i after the loop ends, and why?

Answer : The value of i is 4, because range(5) generates 0, 1, 2, 3, 4, and the loop finishes on 4.
