Q31. Find all Armstrong numbers between 1 and 9999. (An n-digit number is Armstrong if the sum of
its digits each raised to the power n equals the number itself. Example: 153 = 1³ + 5³ + 3³.) Print them
in a clean comma-separated list with a count at the end.


Answer:



```python
count = 0

print("Armstrong numbers:")

for number in range(1, 10000):

    temp = number
    digits = 0

    while temp > 0:
        digits = digits + 1
        temp = temp // 10

    temp = number
    total = 0

    while temp > 0:
        digit = temp % 10
        total = total + digit ** digits
        temp = temp // 10

    if total == number:
        if count > 0:
            print(",", end=" ")

        print(number, end="")
        count = count + 1

print("\nTotal count:", count)
```

Q33. Prime check, optimized. Take a number N. Check if it is prime by iterating only up to √N. Then,
print the SUM of all primes between 2 and N (inclusive). Do not use any library.

Answer:

```python
n = int(input("Enter a number: "))

# Check whether the entered number is prime
is_prime = True

if n < 2:
    is_prime = False

else:
    for i in range(2, n):
        if i * i > n:
            break

        if n % i == 0:
            is_prime = False
            break

if is_prime:
    print(n, "is a prime number")
else:
    print(n, "is not a prime number")


# Find the sum of all prime numbers from 2 to N
prime_sum = 0

for number in range(2, n + 1):

    is_prime = True

    for i in range(2, number):
        if i * i > number:
            break

        if number % i == 0:
            is_prime = False
            break

    if is_prime:
        prime_sum = prime_sum + number

print("Sum of prime numbers:", prime_sum)
```

Q34. Take a number as input. Reverse it WITHOUT converting to string. Then check if the original and
reversed are equal (palindrome). Handle negative numbers, they should be considered NOT
palindrome (because of the minus sign). Examples: 121 → palindrome, 123 → not palindrome, -121

```python
number = int(input("Enter a number: "))

if number < 0:
    print("Reversed number:", number)
    print("Negative numbers are not palindromes")

else:
    original = number
    reversed_number = 0

    while number > 0:
        digit = number % 10
        reversed_number = reversed_number * 10 + digit
        number = number // 10

    print("Reversed number:", reversed_number)

    if original == reversed_number:
        print("Palindrome")
    else:
        print("Not a palindrome")
```

Q35. Take two positive integers. Compute their HCF (Highest Common Factor) using Euclidean
algorithm with a `while` loop. Then compute LCM using the relation: HCF × LCM = a × b. Do not use
math.gcd.

Answer:

```python
a = int(input("Enter first positive number: "))
b = int(input("Enter second positive number: "))

if a <= 0 or b <= 0:
    print("Please enter positive numbers")

else:
    x = a
    y = b

    while y != 0:
        remainder = x % y
        x = y
        y = remainder

    hcf = x
    lcm = (a * b) // hcf

    print("HCF:", hcf)
    print("LCM:", lcm)
```

Q38. Word Frequency Analyzer for Swiggy reviews. Take a review sentence. Find the frequency of
each word (case-insensitive, "Food" and "food" should be counted together). Print words sorted by
frequency in descending order. Also print the MOST frequent word and its count. Use a dict.

Answer:


```python
review = input("Enter a review: ")

words = review.lower().split()

frequency = {}

for word in words:

    if word in frequency:
        frequency[word] = frequency[word] + 1

    else:
        frequency[word] = 1

items = list(frequency.items())

# Arrange words by frequency in descending order
for i in range(len(items)):

    for j in range(i + 1, len(items)):

        if items[j][1] > items[i][1]:
            temp = items[i]
            items[i] = items[j]
            items[j] = temp

print("\nWord frequencies:")

for word, count in items:
    print(word, ":", count)

if len(items) > 0:
    print("\nMost frequent word:", items[0][0])
    print("Count:", items[0][1])
```

Q41. TCS stock prices for a week stored as a tuple: `prices = (3845, 3902, 3878, 3955, 4012, 3998,
4055)` (Mon to Sun in ₹). Find: (a) maximum price and which day, (b) minimum price and which day,
(c) average, (d) biggest single-day gain (and between which days), (e) total volatility = sum of
absolute day-to-day price changes.

Answer:

```python
prices = (3845, 3902, 3878, 3955, 4012, 3998, 4055)

days = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")

maximum = max(prices)
minimum = min(prices)
average = sum(prices) / len(prices)

max_day = days[prices.index(maximum)]
min_day = days[prices.index(minimum)]

biggest_gain = 0
gain_start = ""
gain_end = ""

volatility = 0

for i in range(1, len(prices)):

    change = prices[i] - prices[i - 1]

    if change > biggest_gain:
        biggest_gain = change
        gain_start = days[i - 1]
        gain_end = days[i]

    if change < 0:
        volatility = volatility - change
    else:
        volatility = volatility + change

print("Maximum price:", maximum, "on", max_day)
print("Minimum price:", minimum, "on", min_day)
print("Average price:", round(average, 2))
print("Biggest single-day gain:", biggest_gain)
print("Gain between:", gain_start, "and", gain_end)
print("Total volatility:", volatility)
```

Q45. Caesar Cipher Encryptor. Take a message and a shift value N as input. Encrypt by shifting each
letter forward by N positions in the alphabet (wrapping around, Z + 1 = A). Preserve case. Keep digits,
spaces, and special characters UNCHANGED. Example: "Hello, India!" with shift 3 → "Khoor, Lqgld!"


Answer:

```python
message = input("Enter a message: ")
shift = int(input("Enter shift value: "))

shift = shift % 26

encrypted = ""

for ch in message:

    if ch >= "A" and ch <= "Z":

        position = ord(ch) - ord("A")
        new_position = (position + shift) % 26
        encrypted = encrypted + chr(new_position + ord("A"))

    elif ch >= "a" and ch <= "z":

        position = ord(ch) - ord("a")
        new_position = (position + shift) % 26
        encrypted = encrypted + chr(new_position + ord("a"))

    else:
        encrypted = encrypted + ch

print("Encrypted message:", encrypted)
```

Q46. Write a function `calculate_gst(amount, gst_rate=18, gst_inclusive=False)`:


WhatsApp Image 2026-11-16 at 3.36.44 PM.jpeg

Answer:

```python
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):

    if gst_inclusive == False:

        base_amount = amount
        gst_amount = amount * gst_rate / 100
        total_amount = base_amount + gst_amount

    else:

        total_amount = amount
        base_amount = amount / (1 + gst_rate / 100)
        gst_amount = total_amount - base_amount

    base_amount = round(base_amount, 2)
    gst_amount = round(gst_amount, 2)
    total_amount = round(total_amount, 2)

    return (base_amount, gst_amount, total_amount)


print(calculate_gst(1000))
print(calculate_gst(1180, gst_inclusive=True))
print(calculate_gst(500, gst_rate=5))
```

Q47. Write a RECURSIVE function `power(base, exp)` that computes base raised to exp, WITHOUT
using the `**` operator and WITHOUT using `pow()`. Handle exp = 0 (return 1) and negative exp
(return 1/power(base, -exp)). Test with: power(2, 10), power(5, 0), power(3, -2).


Answer:



```python
def power(base, exp):

    if exp == 0:
        return 1

    elif exp < 0:
        return 1 / power(base, -exp)

    else:
        return base * power(base, exp - 1)


print("power(2, 10) =", power(2, 10))
print("power(5, 0) =", power(5, 0))
print("power(3, -2) =", power(3, -2))
```

Q48. Identity Matrix Manipulations. Using NumPy: (a) create a 5×5 identity matrix using np.eye(), (b)
multiply every element by 10, (c) transpose it (yes, it looks the same, explain in a comment WHY
transpose of an identity is itself), (d) compute the sum of diagonal elements, (e) compute the sum of
all elements. Print each step.


Answer:

```python
import numpy as np

# Create a 5x5 identity matrix
matrix = np.eye(5)

print("Original Matrix:")
print(matrix)

# Multiply every element by 10
matrix = matrix * 10

print("\nAfter multiplying by 10:")
print(matrix)

# Transpose the matrix
transposed = matrix.T

# Identity matrix is symmetric, so its transpose is the same.
print("\nTransposed Matrix:")
print(transposed)

# Sum of diagonal elements
diagonal_sum = np.trace(matrix)

print("\nSum of diagonal elements:", diagonal_sum)

# Sum of all elements
total_sum = np.sum(matrix)

print("Sum of all elements:", total_sum)
```

Q49. Zomato Weekly Sales Analytics. Create a NumPy array of 7-day Zomato Bengaluru sales (in ₹):

WhatsApp Image 2026-11-16 at 4.11.50 PM.jpeg



Without using loops, compute and print:





WhatsApp Image 2026-11-16 at 4.11.50 PM (1).jpeg




Answer:

```python
import numpy as np

sales = np.array([12000, 18500, 14200, 22100, 19800, 35200, 28400])

days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])

# Total weekly sales
total_sales = np.sum(sales)

# Daily average
average = np.mean(sales)

# Day with maximum sales
maximum_index = np.argmax(sales)
maximum_day = days[maximum_index]

# Number of days above average
above_average = np.sum(sales > average)

# Percentage growth from Monday to Sunday
growth = ((sales[6] - sales[0]) / sales[0]) * 100

# Day-on-day percentage changes
daily_changes = ((sales[1:] - sales[:-1]) / sales[:-1]) * 100

print("Total weekly sales:", total_sales)
print("Daily average:", round(average, 2))
print("Day with maximum sales:", maximum_day)
print("Days above average:", above_average)
print("Monday to Sunday growth:", round(growth, 2), "%")
print("Day-on-day percentage changes:")
print(np.round(daily_changes, 2))
```

Q50. The Border Pattern Challenge – NO explicit loops. Create a 5×5 NumPy array where the BORDER
is 1 and the INTERIOR is 0. Output should look like:

WhatsApp Image 2026-11-16 at 4.11.51 PM.jpeg

Answer:

```python
import numpy as np

# Create a 5x5 matrix filled with 1
matrix = np.ones((5, 5), dtype=int)

# Make the inner area 0
matrix[1:4, 1:4] = 0

print(matrix)
```
